````python
def init_kernel(in_channels, out_channels, kernel_size):
    w = np.random.randn(
        in_channels,
        out_channels,
        kernel_size,
        kernel_size
    )
    bias = np.zeros(out_channels)
    
    #test w
    # w = np.ones((1, 8, 3, 3))
    # bias = np.zeros(8)

    return w, bias

def init_linear_w(in_channels, out_channels):
    w = np.random.rand(
            out_channels, 
            in_channels
        )
    bias = np.zeros(out_channels)
    return w, bias
````

````python
def convolutional(data, weigh, bias, padding=1, stride=1):
    original_size = data.shape[1]
    in_channels = weigh.shape[0]
    out_channels = weigh.shape[1]
    kernel_size = weigh.shape[2]
    
    pad_width = ((0,0), (padding, padding), (padding, padding))
    data = np.pad(data, pad_width, mode='constant', )
    
    out_size = int((original_size - kernel_size + 2 * padding) / stride + 1)

    res = np.zeros((out_channels, out_size, out_size))
    
    for feat_map_idx, feature_map in enumerate(data):
        points = np.zeros((out_size, out_size))
        Filter = weigh[:, feat_map_idx, :, :]
        for i in range(out_size):
            for j in range(out_size):
                input_patch = data[:, i:i+kernel_size, j:j+kernel_size]
                product = np.sum(input_patch * Filter) + bias[feat_map_idx]
                points[i][j] = product
        res[feat_map_idx] = points
    return res
````

````python
def conv2D(data, w, bias, padding=1, stride=1):
    res = convolutional(data, w, bias, padding=padding, stride=stride)
    return res
````

````python
def Relu(data):
    res = np.where(data > 0, data, 0)
    return res
````

````python
def MaxPool(data, size):
    features = data.shape[0]
    data_size = data.shape[1]
    
    out_size = int((data.shape[1] - size) / size + 1)
    res = np.zeros((features, out_size, out_size))
    
    for idx, feature_map in enumerate(data):
        submatrices = []
        for i in range(0, data_size, size):
            for j in range(0, data_size, size):
                sub_arr = np.array(feature_map[i:i+size, j:j+size])
                if sub_arr.shape[0] == size and sub_arr.shape[1] == size:
                    maxpool = np.max(sub_arr)
                    submatrices.append(maxpool)
        submatrices = np.array(submatrices)
        submatrices = submatrices.reshape(out_size, out_size)
                    
        res[idx] = submatrices
    return res   
````

````python
def flatten(data):
    res = []
    
    for feat in data:
        for i in range(len(feat)):
            for j in range(len(feat)):
                res.append(feat[i][j])
    return np.array(res)
````

````python
def Linear(data, w, bias):
    res = []
    
    for idx, feat in enumerate(w):
        tmp = data @ feat
        tmp += bias[idx]
        res.append(tmp)
    return np.array(res), data
````

````python
def Softmax(data):
    sum = 0
    res = []
    for vals in data:
        sum += np.exp(vals)
    for vals in data:
        res.append(np.exp(vals)/sum)
    return res
        
````

````python
def pick_number(p):
    res_vals = 0
    pred = 0
    for idx, vals in enumerate(p):
        if (vals > res_vals):
            pred = idx 
            res_vals = vals 
    return pred
````

````python
def Loss(label, pred):
    loss = 0

    for i in range(len(label)):
        loss += -label[i] * np.log(pred[i])

    return loss
````

````python
cache = {}

W1, b1 = init_kernel(1, 8, 3)
W2, b2 = init_kernel(8, 16, 3)

W3, b3 = init_linear_w(16, 10)

def Model(inputs):
    cache["conv1_input"] = inputs
    outputs = conv2D(inputs, W1, b1)
    cache["conv1_W"] = W1
    cache["conv1_bias"] = b1
    cache["relu1_input"] = outputs
    outputs = Relu(outputs)
    cache["pool1_input"] = outputs
    outputs = MaxPool(outputs, 2)
    cache["conv2_input"] = outputs

    outputs = conv2D(outputs, W2, b2)
    cache["conv2_W"] = W2
    cache["conv2_bias"] = b2
    cache["relu2_input"] = outputs
    outputs = Relu(outputs)
    cache["pool2_input"] = outputs
    outputs = MaxPool(outputs, 2)


    cache["flatten_shape"] = outputs.shape
    outputs = flatten(outputs)

    outputs, linear_data = Linear(outputs, W3, b3)
    cache["linear_data"] = linear_data
    cache["linear_W"] = W3
    cache["linear_bias"] = b3
    
    outputs = Softmax(outputs)

    return outputs
````

````python
def Softmax_backward(pred, target):
    return pred - target

grad = Softmax_backward(p, label)
````

````python
def Linear_backward(grad, data, w):

    dW = np.outer(grad, data)

    db = grad.copy()

    dx = w.T @ grad

    return dx, dW, db

data = cache["linear_data"]
W = cache["linear_W"]
dx, dW, db = Linear_backward(grad, data, W)
print("dx.shape: ", dx.shape)
print("dW.shape: ", dW.shape)
print("db.shape: ", db.shape)
````

````python
def flatten_backward(dx, original_shape):
    return dx.reshape(original_shape)

original_shape = cache["flatten_shape"]
dx = flatten_backward(dx, original_shape)

print(dx.shape) 
````

````python
def MaxPool_backward(dout, data, size):

    dx = np.zeros_like(data)

    C, H, W = data.shape

    out_H = H // size
    out_W = W // size

    for c in range(C):
        for i in range(out_H):
            for j in range(out_W):

                # Lấy vùng tương ứng với output
                patch = data[
                    c,
                    i*size:(i+1)*size,
                    j*size:(j+1)*size
                ]

                # Tìm vị trí max
                max_idx = np.unravel_index(
                    np.argmax(patch),
                    patch.shape
                )

                # Đưa gradient về vị trí max
                dx[
                    c,
                    i*size + max_idx[0],
                    j*size + max_idx[1]
                ] = dout[c, i, j]

    return dx

dx = MaxPool_backward(
    dx,
    cache["pool2_input"],
    2
)
print(dx.shape)
````

````python
def Relu_backward(dout, data):

    dx = dout.copy()

    dx[data <= 0] = 0

    return dx

dx = Relu_backward(dx, cache["relu2_input"])
print(dx.shape)
````

````python
def conv2D_backward(dout, data, W, bias):

    # data: (in_channels, H, W)
    in_channels, H, W_input = data.shape

    # W: (out_channels, in_channels, kernel_size, kernel_size)
    out_channels, _, kernel_size, _ = W.shape

    H_out = H - kernel_size + 1
    W_out = W_input - kernel_size + 1

    dx = np.zeros_like(data)
    dW = np.zeros_like(W)
    db = np.zeros_like(bias)

    for co in range(out_channels):

        for i in range(H_out):

            for j in range(W_out):

                db[co] += dout[co, i, j]

                for ci in range(in_channels):

                    for u in range(kernel_size):

                        for v in range(kernel_size):

                            x = data[ci, i + u, j + v]
                            w = W[co, ci, u, v]

                            dW[co, ci, u, v] += dout[co, i, j] * x

                            dx[ci, i + u, j + v] += dout[co, i, j] * w

    return dx, dW, db
````

````python
# print(grad.shape)
def Backward(grad):
    dx, dW, db = Linear_backward(
        grad,
        cache["linear_data"],
        cache["linear_W"]
    )
    # print(dx.shape)
    dx = flatten_backward(
        dx,
        cache["flatten_shape"]
    )
    # print(dx.shape)
    
    dx = MaxPool_backward(
        dx,
        cache["pool2_input"],
        2
    )
    # print(dx.shape)
    
    dx = Relu_backward(
        dx,
        cache["relu2_input"]
    )
    # print(dx.shape)
    
    dx, dW2, db2 = conv2D_backward(
        dx,
        cache["conv2_input"],
        cache["conv2_W"],
        cache["conv2_bias"]
    )
    # print(dx.shape)
    
    #=================================
    dx = MaxPool_backward(
        dx,
        cache["pool1_input"],
        2
    )
    # print(dx.shape)
    
    dx = Relu_backward(
        dx,
        cache["relu1_input"]
    )
    # print(dx.shape)
    
    dx, dW1, db1 = conv2D_backward(
        dx,
        cache["conv1_input"],
        cache["conv1_W"],
        cache["conv1_bias"]
    )
    # print(dx.shape)

    return dW1, db1, dW2, db2, dW, db
````

````python
def update(lr):
    global W1, b1, W2, b2, W3, b3
    dW1, db1, dW2, db2, dW3, db3 = Backward(grad)

    W1 -= lr * dW1
    b1 -= lr * db1

    W2 -= lr * dW2
    b2 -= lr * db2

    W3 -= lr * dW3
    b3 -= lr * db3
    
update(0.1)

print(W1, b1)
print(W2, b2)
print(W3, b3)
````
